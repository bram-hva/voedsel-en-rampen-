```python
import pandas as pd

fao_totaal = pd.read_csv("data/fao_granen_totaal.csv")
fao_tmr = pd.read_csv("data/fao_tarwe_mais_rijst.csv")
fao_sgg = pd.read_csv("data/fao_sorghum_gierst_gerst.csv")

fao = pd.concat([fao_totaal, fao_tmr, fao_sgg], ignore_index=True)

print(fao.shape)
print(fao.columns.tolist())
fao.head()
```

```python
rampen = pd.read_csv("data/emdat_rampen_1970_2021.csv")

print(rampen.shape)
print(rampen.columns.tolist())
print(rampen["Disaster Type"].value_counts())
rampen.head()
```

```python
fao_landen = set(fao["Area"])
ramp_landen = set(rampen["Country"])

niet_gevonden = ramp_landen - fao_landen
print(len(ramp_landen), "landen in EM-DAT")
print(len(niet_gevonden), "daarvan niet terug te vinden in FAO")
print(sorted(niet_gevonden))
```

```python
landcodes = pd.read_csv("data/fao_landcodes.csv")

print(landcodes.shape)
print(landcodes.columns.tolist())
landcodes.head()
```

```python
# Alleen de twee kolommen die we nodig hebben uit de vertaaltabel
landcodes = landcodes[["M49 Code", "ISO3 Code"]]

print(fao["Area Code (M49)"].dtype, landcodes["M49 Code"].dtype)

fao = fao.merge(landcodes, left_on="Area Code (M49)", right_on="M49 Code",
                how="left", validate="many_to_one")

print(fao.shape)
print(fao["ISO3 Code"].isna().sum(), "rijen zonder ISO3-code")
```

```python
fao_codes = set(fao["ISO3 Code"])
ramp_codes = set(rampen["ISO"])

niet_gevonden = ramp_codes - fao_codes
print(len(ramp_codes), "landen in EM-DAT")
print(len(niet_gevonden), "daarvan niet terug te vinden in FAO")

# Welke landen zijn dat? Toon de naam erbij.
print(rampen[rampen["ISO"].isin(niet_gevonden)]["Country"].unique())
```

```python
productie = fao[["ISO3 Code", "Area", "Item", "Year", "Value"]]
productie = productie.rename(columns={
    "ISO3 Code": "ISO3",
    "Area": "Land",
    "Item": "Gewas",
    "Year": "Jaar",
    "Value": "Ton",
})

print(productie.shape)
print(productie["Gewas"].unique())
print(productie["Ton"].isna().sum(), "rijen zonder productiecijfer")

# Controle: hoe staat China erin?
print(productie[productie["Land"].str.contains("China")]["Land"].unique())
```

```python
# Welke ISO3-code heeft elke China-variant?
china = productie[productie["Land"].str.contains("China")]
print(china[["Land", "ISO3"]].drop_duplicates())

# Rijen zonder productiecijfer weggooien (niet op 0 zetten!)
productie = productie.dropna(subset=["Ton"])
print(productie.shape)
```

```python
rampen_schoon = rampen[["ISO", "Country", "Continent", "Year", "Start Year",
                        "Disaster Type", "Total Affected"]]
rampen_schoon = rampen_schoon.rename(columns={
    "ISO": "ISO3",
    "Country": "Land",
    "Continent": "Werelddeel",
    "Disaster Type": "Ramptype",
    "Total Affected": "Getroffen",
})

# Zijn "Year" en "Start Year" altijd hetzelfde?
print((rampen_schoon["Year"] != rampen_schoon["Start Year"]).sum(), "rijen waar ze verschillen")

# Hoe vaak heeft één land meerdere droogtes in hetzelfde jaar?
droogtes = rampen_schoon[rampen_schoon["Ramptype"] == "Drought"]
print(len(droogtes), "droogtes")
print(droogtes.duplicated(subset=["ISO3", "Start Year"]).sum(), "dubbele land-jaar-combinaties")
```

```python
# Start Year wordt ons Jaar; de kolom Year hebben we niet meer nodig
rampen_schoon = rampen_schoon.drop(columns=["Year"])
rampen_schoon = rampen_schoon.rename(columns={"Start Year": "Jaar"})

# Per land, jaar en soort ramp: hoeveel rampen waren het?
rampjaren = rampen_schoon.groupby(["ISO3", "Jaar", "Ramptype"]).size().reset_index(name="Aantal_rampen")

print(rampjaren.shape)
print(len(rampjaren[rampjaren["Ramptype"] == "Drought"]), "droogte-jaren")
rampjaren.head()
```

```python
breed = productie.pivot(index=["ISO3", "Gewas"], columns="Jaar", values="Ton")

print(breed.shape)
print(breed.columns.min(), "t/m", breed.columns.max())

# Bekijk één voorbeeld: Ethiopië, granen totaal, rond de droogte van 1984
breed.loc[("ETH", "Cereals, primary"), 1982:1986]
```

```python
# In welke jaren staat Ethiopië (of een variant) in de FAO-data, en met welke code?
ethiopie = productie[productie["Land"].str.contains("Ethiopia")]
print(ethiopie.groupby(["Land", "ISO3"])["Jaar"].agg(["min", "max"]))

# En Soedan?
soedan = productie[productie["Land"].str.contains("Sudan")]
print(soedan.groupby(["Land", "ISO3"])["Jaar"].agg(["min", "max"]))
```

```python
# Oude FAO-landen doorzetten onder de code van het huidige land
oude_codes = {"F62": "ETH", "F206": "SDN"}
productie["ISO3"] = productie["ISO3"].replace(oude_codes)

breed = productie.pivot(index=["ISO3", "Gewas"], columns="Jaar", values="Ton")
print(breed.shape)
breed.loc[("ETH", "Cereals, primary"), 1982:1986]
```

```python
jaar_ervoor = breed.shift(1, axis=1)
jaar_erna = breed.shift(-1, axis=1)

# De vaste definitie uit de opdracht: van het jaar ervóór naar het jaar ná, in procenten
verandering = (jaar_erna - jaar_ervoor) / jaar_ervoor * 100

verandering.loc[("ETH", "Cereals, primary"), 1982:1986].round(1)
```

```python
# Wanneer registreerde EM-DAT droogtes in Ethiopië?
print(rampjaren[(rampjaren["ISO3"] == "ETH") & (rampjaren["Ramptype"] == "Drought")]["Jaar"].tolist())

# Van breed terug naar lang
verandering_lang = verandering.reset_index().melt(
    id_vars=["ISO3", "Gewas"], var_name="Jaar", value_name="Verandering_pct")

print(verandering_lang.shape)
print(verandering_lang["Verandering_pct"].isna().sum(), "rijen zonder verandering")
verandering_lang.head()
```

```python
# Lege veranderingen weg, en delen door nul (inf) ook weg
verandering_lang = verandering_lang.replace([float("inf"), float("-inf")], float("nan"))
verandering_lang = verandering_lang.dropna(subset=["Verandering_pct"])

# Alleen de jaren waarin EM-DAT meetelt
verandering_lang = verandering_lang[verandering_lang["Jaar"].between(1970, 2021)]

# Alleen de droogtes uit de rampentabel
droogtejaren = rampjaren[rampjaren["Ramptype"] == "Drought"][["ISO3", "Jaar", "Aantal_rampen"]]

# Koppelen op land + jaar
analyse = verandering_lang.merge(droogtejaren, on=["ISO3", "Jaar"],
                                 how="left", validate="many_to_one")
analyse["Droogtejaar"] = analyse["Aantal_rampen"].notna()

print(analyse.shape)
print(analyse["Droogtejaar"].value_counts())
```

```python
granen = analyse[analyse["Gewas"] == "Cereals, primary"]

print(granen.groupby("Droogtejaar")["Verandering_pct"].agg(["count", "mean", "median"]).round(1))
```

```python
# De klap in het rampjaar zelf: van het jaar ervóór naar het rampjaar
klap = (breed - jaar_ervoor) / jaar_ervoor * 100

klap_lang = klap.reset_index().melt(
    id_vars=["ISO3", "Gewas"], var_name="Jaar", value_name="Klap_pct")

# Naast de bestaande kolommen hangen
analyse = analyse.merge(klap_lang, on=["ISO3", "Gewas", "Jaar"], how="left")

granen = analyse[analyse["Gewas"] == "Cereals, primary"]
print(granen.groupby("Droogtejaar")[["Klap_pct", "Verandering_pct"]].median().round(1))
```

```python
from scipy import stats

droogte = granen[granen["Droogtejaar"]]["Klap_pct"].dropna()
normaal = granen[~granen["Droogtejaar"]]["Klap_pct"].dropna()

print(len(droogte), "droogtejaren,", len(normaal), "andere jaren")
print("gemiddelde:", round(droogte.mean(), 1), "tegen", round(normaal.mean(), 1))

toets = stats.ttest_ind(droogte, normaal, equal_var=False)
interval = toets.confidence_interval()

print("verschil:", round(droogte.mean() - normaal.mean(), 1), "procentpunt")
print("95%-interval:", round(interval.low, 1), "tot", round(interval.high, 1))
print("p-waarde:", round(toets.pvalue, 4))
```

```python
# De productie van het jaar ervoor ook naar lang formaat en erbij hangen
ervoor_lang = jaar_ervoor.reset_index().melt(
    id_vars=["ISO3", "Gewas"], var_name="Jaar", value_name="Ton_ervoor")
analyse = analyse.merge(ervoor_lang, on=["ISO3", "Gewas", "Jaar"], how="left")

granen = analyse[analyse["Gewas"] == "Cereals, primary"]

# De vijf grootste stijgingen: wie zijn dat?
print(granen.nlargest(5, "Klap_pct")[["ISO3", "Jaar", "Ton_ervoor", "Klap_pct"]])

# Hoeveel waarnemingen komen van landen met minder dan 100.000 ton graan?
print((granen["Ton_ervoor"] < 100_000).sum(), "van de", len(granen))
```

```python
def vergelijk(data, kolom, minimale_productie):
    selectie = data[data["Ton_ervoor"] >= minimale_productie]
    droogte = selectie[selectie["Droogtejaar"]][kolom].dropna()
    normaal = selectie[~selectie["Droogtejaar"]][kolom].dropna()

    toets = stats.ttest_ind(droogte, normaal, equal_var=False)
    interval = toets.confidence_interval()
    verschil = droogte.mean() - normaal.mean()

    print(f"drempel {minimale_productie:>9} ton | n = {len(droogte):>3} tegen {len(normaal):>4} | "
          f"verschil {verschil:6.1f} pp  (95%: {interval.low:6.1f} tot {interval.high:6.1f}) | "
          f"p = {toets.pvalue:.4f}")

for drempel in [0, 10_000, 100_000, 1_000_000]:
    vergelijk(granen, "Klap_pct", drempel)
```

```python
# 1. Is het verschil na een jaar echt weg? Zelfde toets, andere kolom.
vergelijk(granen, "Verandering_pct", 100_000)

# 2. De vervolgvraag uit de opdracht: welke landen herstellen binnen een jaar?
droogte_granen = granen[granen["Droogtejaar"] & (granen["Ton_ervoor"] >= 100_000)].copy()
droogte_granen["Hersteld"] = droogte_granen["Verandering_pct"] >= 0

print(droogte_granen["Hersteld"].mean().round(2), "van de droogtes: binnen een jaar hersteld")

per_land = droogte_granen.groupby("ISO3").agg(
    aantal_droogtes=("Hersteld", "size"),
    aandeel_hersteld=("Hersteld", "mean"),
    mediane_klap=("Klap_pct", "median"),
)
per_land = per_land[per_land["aantal_droogtes"] >= 3]
print(per_land.sort_values("aandeel_hersteld").head(10).round(2))
```

```python
## Bevindingen
# - Een droogte kost in het rampjaar zelf 7,5 procentpunt graangroei (95%-interval 5,4 tot 9,6, landen met minstens 100.000 ton).
# - Een jaar later is dat verschil niet meer aantoonbaar: 61% van de landen is dan hersteld.
# - Het herstel hangt niet alleen af van rijkdom: ook Italië, Spanje en Rusland herstellen vaak niet binnen een jaar.
```
